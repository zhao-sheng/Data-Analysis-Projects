```python
# %% [markdown]
# # 1. 电信数据模拟生成
# 生成符合电信业务场景的合成数据

# %%
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

# %%
def generate_towers(n=50):
    """生成基站地理数据"""
    towers = pd.DataFrame({
        'tower_id': [f"T{1000+i}" for i in range(n)],
        'lat': np.random.uniform(48.1, 48.3, n),
        'lon': np.random.uniform(16.2, 16.5, n),
        'frequency_band': np.random.choice(["800MHz", "1.8GHz", "2.6GHz"], n)
    })
    towers.to_csv('../data/synthetic_towers.csv', index=False)
    return towers

# %%
def generate_connections(towers, days=30):
    """生成用户连接时间序列数据"""
    timestamps = pd.date_range(
        start=datetime.now() - timedelta(days=days),
        end=datetime.now(),
        freq='min'
    )
    connections = pd.DataFrame({
        'timestamp': timestamps,
        'tower_id': np.random.choice(towers['tower_id'], len(timestamps)),
        'user_id': [f"U{np.random.randint(10000,99999)}" for _ in range(len(timestamps))],
        'signal_strength': np.random.normal(-75, 10, len(timestamps))
    })
    connections.to_parquet('../data/synthetic_connections.parquet')
    return connections

# %%
# 执行生成
towers = generate_towers()
connections = generate_connections(towers)
```
