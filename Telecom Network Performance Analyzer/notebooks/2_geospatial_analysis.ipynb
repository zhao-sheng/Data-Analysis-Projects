```python
# %% [markdown]
# # 2. 基站覆盖热力图分析
# 使用Folium进行地理空间可视化

# %%
import folium
from folium.plugins import HeatMap
import pandas as pd

# %%
# 加载数据
towers = pd.read_csv('../data/synthetic_towers.csv')
connections = pd.read_parquet('../data/synthetic_connections.parquet')

# %%
def plot_coverage(towers, connections):
    # 计算连接密度
    density = connections['tower_id'].value_counts().reset_index()
    density.columns = ['tower_id', 'connection_count']
    towers = towers.merge(density, on='tower_id', how='left')
    
    # 创建热力图
    m = folium.Map(location=[48.2, 16.3], zoom_start=12)
    HeatMap(
        data=towers[['lat', 'lon', 'connection_count']].values,
        radius=15
    ).add_to(m)
    
    # 保存交互式结果
    m.save('../reports/coverage_heatmap.html')
    return m

# %%
# 生成可视化
map_obj = plot_coverage(towers, connections)
map_obj
```
