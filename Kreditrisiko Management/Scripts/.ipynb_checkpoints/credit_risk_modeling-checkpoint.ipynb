```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, roc_auc_score
from imblearn.over_sampling import SMOTE
import xgboost as xgb
from pathlib import Path

# Set plotting style
plt.style.use('ggplot')
sns.set_palette('muted')

# Define paths
DATA_DIR = Path('../Data')
PLOTS_DIR = Path('../Plots')

# Ensure Plots directory exists
PLOTS_DIR.mkdir(exist_ok=True)

# 1. Load the cleaned dataset
df = pd.read_csv(DATA_DIR / 'accepted_cleaned.csv')
print("=== Dataset Overview ===")
print(df.info())
print("\n=== First 5 Rows ===")
print(df.head())

# Check class distribution to identify imbalance
print("\n=== Class Distribution Before Modeling ===")
print(df['default_status'].value_counts(normalize=True))
print(df['default_status'].value_counts())

# 2. Feature Engineering
# Select features and target
features = ['loan_amnt', 'int_rate', 'grade', 'annual_inc', 'dti', 'fico_range_low', 'delinq_2yrs', 'open_acc', 'revol_bal']
target = 'default_status'

# Add a new feature: loan-to-income ratio
df['loan_to_income'] = df['loan_amnt'] / df['annual_inc']
features.append('loan_to_income')

# Prepare X (features) and y (target)
X = df[features]
y = df[target]

# 3. Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 4. Apply SMOTE to balance the training set
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("\n=== Class Distribution After SMOTE (Training Set) ===")
print(pd.Series(y_train_smote).value_counts(normalize=True))
print(pd.Series(y_train_smote).value_counts())

# 5. Standardize the features
scaler = StandardScaler()
X_train_smote_scaled = scaler.fit_transform(X_train_smote)
X_test_scaled = scaler.transform(X_test)

# 6. Train a Logistic Regression model
log_model = LogisticRegression(random_state=42)
log_model.fit(X_train_smote_scaled, y_train_smote)

# 7. Train an XGBoost model
xgb_model = xgb.XGBClassifier(
    random_state=42,
    scale_pos_weight=1,
    use_label_encoder=False,
    eval_metric='logloss'
)
xgb_model.fit(X_train_smote_scaled, y_train_smote)

# 8. Evaluate both models
# Function to evaluate model
def evaluate_model(model, X_test_scaled, y_test, threshold=0.5, model_name="Model"):
    y_pred_prob = model.predict_proba(X_test_scaled)[:, 1]
    y_pred = (y_pred_prob >= threshold).astype(int)

    print(f"\n=== {model_name} Classification Report (Threshold = {threshold}) ===")
    print(classification_report(y_test, y_pred))

    print(f"\n=== {model_name} Confusion Matrix (Threshold = {threshold}) ===")
    conf_matrix = confusion_matrix(y_test, y_pred)
    print(conf_matrix)

    roc_auc = roc_auc_score(y_test, y_pred_prob)
    print(f"\n=== {model_name} ROC-AUC Score: {roc_auc:.4f} ===")

    return y_pred_prob, conf_matrix, roc_auc

# Evaluate Logistic Regression with threshold 0.5
y_pred_prob_log, conf_matrix_log, roc_auc_log = evaluate_model(
    log_model, X_test_scaled, y_test, threshold=0.5, model_name="Logistic Regression"
)

# Evaluate XGBoost with threshold 0.5
y_pred_prob_xgb, conf_matrix_xgb, roc_auc_xgb = evaluate_model(
    xgb_model, X_test_scaled, y_test, threshold=0.5, model_name="XGBoost"
)

# 9. Save predictions from the better model (XGBoost)
pd.DataFrame(y_test, columns=['default_status']).to_csv(DATA_DIR / 'y_test.csv', index=False)
pd.DataFrame(y_pred_prob_xgb, columns=['pred_prob']).to_csv(DATA_DIR / 'y_pred_prob.csv', index=False)
print(f"Saved {DATA_DIR / 'y_test.csv'} and {DATA_DIR / 'y_pred_prob.csv'}")

# 10. Visualize the results (using XGBoost)
# (1) Confusion Matrix Heatmap
plt.figure(figsize=(10, 8))
# Transpose the confusion matrix to match Dash's orientation
conf_matrix_xgb = conf_matrix_xgb.T
annot_colors = np.array([['black', 'black'], ['black', 'black']])
for i in range(2):
    for j in range(2):
        if conf_matrix_xgb[i, j] > 1000:
            annot_colors[i, j] = 'white'
sns.heatmap(conf_matrix_xgb, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['No Default', 'Default'],  # Actual labels
            yticklabels=['No Default', 'Default'],  # Predicted labels
            annot_kws={'size': 12})
for i in range(2):
    for j in range(2):
        plt.text(j + 0.5, i + 0.5, conf_matrix_xgb[i, j],
                 ha='center', va='center', color=annot_colors[i, j], size=12)
plt.title('XGBoost Confusion Matrix (Threshold = 0.5)')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.tight_layout()
plt.savefig(PLOTS_DIR / 'confusion_matrix.png', dpi=300)
plt.close()

# (2) ROC Curve
plt.figure(figsize=(10, 8))
fpr, tpr, _ = roc_curve(y_test, y_pred_prob_xgb)
plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {roc_auc_xgb:.2f})', color='blue', linewidth=2)
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess', linewidth=1.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.title('XGBoost Receiver Operating Characteristic (ROC) Curve', fontsize=14)
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.legend(loc='lower right', fontsize=12)
plt.tick_params(axis='both', labelsize=10)
plt.tight_layout()
plt.savefig(PLOTS_DIR / 'roc_curve.png', dpi=300)
plt.close()

# (3) Predicted Probability Distribution
plt.figure(figsize=(10, 8))
sns.histplot(y_pred_prob_xgb, bins=50, kde=True, color='lightblue', stat='count')
sns.kdeplot(y_pred_prob_xgb, color='blue', linewidth=2)
plt.axvline(x=0.5, color='red', linestyle='--', label='Threshold = 0.5', linewidth=1.5)
plt.title('XGBoost Distribution of Predicted Default Probabilities', fontsize=14)
plt.xlabel('Predicted Probability of Default', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend(fontsize=12)
plt.tick_params(axis='both', labelsize=10)
plt.tight_layout()
plt.savefig(PLOTS_DIR / 'predicted_probability_distribution.png', dpi=300)
plt.close()

# 11. Feature Importance (XGBoost)
plt.figure(figsize=(10, 8))
feat_importances = pd.Series(xgb_model.feature_importances_, index=features)
feat_importances.nlargest(10).plot(kind='barh')
plt.title('XGBoost Feature Importance', fontsize=14)
plt.xlabel('Importance', fontsize=12)
plt.ylabel('Feature', fontsize=12)
plt.tight_layout()
plt.savefig(PLOTS_DIR / 'feature_importance.png', dpi=300)
plt.close()

# 12. Save the model coefficients (Logistic Regression) and feature importances (XGBoost)
coef_df = pd.DataFrame({
    'Feature': features,
    'Coefficient': log_model.coef_[0]
})
print("\n=== Logistic Regression Coefficients ===")
print(coef_df)
coef_df.to_csv(DATA_DIR / 'model_coefficients.csv', index=False)
print(f"Saved {DATA_DIR / 'model_coefficients.csv'}")

feat_imp_df = pd.DataFrame({
    'Feature': features,
    'Importance': xgb_model.feature_importances_
})
print("\n=== XGBoost Feature Importances ===")
print(feat_imp_df)
feat_imp_df.to_csv(DATA_DIR / 'feature_importances.csv', index=False)
print(f"Saved {DATA_DIR / 'feature_importances.csv'}")
```
